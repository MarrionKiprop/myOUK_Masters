## 2.0 Implementing Count Sort with Radix Sort in Python 1


```python
"""   
Implement Count Sort as an helper for the Radix sort  
"""
# Counting Sort as helper function
def counting_sort(arr, place):
    size = len(arr)
    output = [0] * size
    count = [0] * 10  # for digits 0–9

    # Count occurrences of each digit
    for num in arr:
        digit = (num // place) % 10
        count[digit] += 1

    # Cumulative count
    for i in range(1, 10):
        count[i] += count[i - 1]

    # Build output array (stable sorting)
    i = size - 1
    while i >= 0:
        digit = (arr[i] // place) % 10
        output[count[digit] - 1] = arr[i]
        count[digit] -= 1
        i -= 1

    # Copy sorted data back into original array
    for i in range(size):
        arr[i] = output[i]


# Radix Sort implementation using provided phone numbers
def radix_sort(arr):
    """
    Implementing Radix Sort on the provided phone numbers
    """
    max_num = max(arr)
    place = 1
    step = 1

    while max_num // place > 0:
        print(f"\nStep {step}: Sorting by place = {place} (digit position)")
        counting_sort(arr, place)
        print(f"After sorting by {place}'s place: {arr}")
        place *= 10
        step += 1

    return arr

# Example: list of phone numbers
phone_numbers = [564, 213, 987, 432, 123, 765, 321, 654, 876]

print("Original phone numbers:", phone_numbers)
sorted_numbers = radix_sort(phone_numbers)
print("\nFinal sorted phone numbers:", sorted_numbers)
```

## 2.1 Radix Sort on Provided Phone Numbers

In the first step, the numbers are sorted using Counting Sort based on the last digit (LSD). This rearranges the list so that numbers ending in 1 come before those ending in higher digits like 7. This forms the first intermediate order.

The second step sorts the updated list using the middle digit. Because the sort is stable, the order from the LSD step is preserved. Numbers whose middle digit is 1 appear first, followed by those with middle digit 2, and so on.

The final step sorts based on the first digit (MSD). This gives the complete ascending order, with numbers starting with 1 appearing before those starting with 2, 3, etc.

Since all numbers have the same length, Radix Sort processes them smoothly, one digit at a time, from LSD to MSD.


## 3.0 Implementing Count Sort with Radix Sort in Python 2

```python
# Radix Sort implementation on different phone numbers of different lengths
def radix_sort(arr):
    """
    Implementing Radix Sort using different length phone numbers from a 
    single digit to six digit phone number
    """
    max_num = max(arr)
    place = 1
    step = 1

    while max_num // place > 0:
        print(f"\nStep {step}: Sorting by place = {place} (digit position)")
        counting_sort(arr, place)
        print(f"After sorting by {place}'s place: {arr}")
        place *= 10
        step += 1

    return arr

# My phone numbers
phone_numbers = [7, 45, 123, 8901, 56, 9, 320, 78124, 604, 28, 5, 1002]

print("Different Length phone numbers:", phone_numbers)
sorted_numbers = radix_sort(phone_numbers)
print("\nFinal sorted phone numbers of different lengths:", sorted_numbers)
```

## 3.1 Radix Sort on Phone Numbers with Different Lengths

In this case, the phone numbers have different lengths, so Radix Sort performs several passes—one for each digit position up to the longest number. The process still begins with the LSD, where all numbers are sorted by their last digit, regardless of how many digits they have. On the next passes, shorter numbers behave as if they have leading zeros, so once their actual digits are used, they remain in the correct position. This continues until the MSD, where the longest numbers finally sort fully. Even with mixed lengths, Radix Sort still works correctly because each pass is stable and only considers one digit at a time.
