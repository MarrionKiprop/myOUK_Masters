## 1. Problem Analysis

### 1.1 Formal Definition of Subset Sum

Given:

- A set of integers  
  S = { x1, x2, x3, ..., xn }

- A target integer  
  T

Decision version (YES/NO question):

Does there exist a subset of S whose sum equals T?

Formally, determine whether there exists a subset  
S' ⊆ S  
such that:

sum_{x in S'} x = T

That is, the sum of all elements in S' is exactly equal to T.


### 1.2 Why Subset Sum Is NP-Complete

Subset Sum is NP-complete because it satisfies two conditions:

1. Subset Sum is in NP

Given a candidate subset S' of S, we can verify in polynomial time whether it is a valid solution.

Verification process:
- Compute the sum of all elements in S'.
- Check if this sum is equal to the target T.
- If sum(S') == T, then S' is a valid solution.

This verification requires only a linear pass over S', plus a comparison, which is polynomial in the size of the input. Therefore, Subset Sum is in NP.

2. Subset Sum is NP-hard

There exists a classical polynomial-time reduction from a known NP-complete problem (such as 3-SAT) to Subset Sum (as shown in Karp’s list of 21 NP-complete problems).

This means:
- Any instance of 3-SAT can be transformed into an equivalent instance of Subset Sum in polynomial time.
- If we could solve Subset Sum efficiently (in polynomial time), then we could also solve 3-SAT efficiently.
- Since 3-SAT is NP-complete, solving it efficiently would imply we can solve all problems in NP efficiently.

Thus, Subset Sum is at least as hard as the hardest problems in NP.

Conclusion:

- Subset Sum is in NP (its solutions can be verified in polynomial time), and  
- Subset Sum is NP-hard (every problem in NP can be reduced to it).

Therefore:

Subset Sum is NP-complete.


### 1.3 Real-World Applications

Subset Sum appears in many real-life optimization and security settings, including:

1. Cryptography  
   Some knapsack-based public-key cryptosystems (such as the Merkle–Hellman scheme) are based on the hardness of Subset Sum.

2. Resource allocation  
   Selecting a combination of items (e.g., projects, tasks, or goods) whose total cost or resource use matches a given budget exactly.

3. Load balancing  
   Assigning tasks to processors or machines so that one subset of tasks matches a specific load target.

4. Financial portfolio selection  
   Choosing a subset of assets whose combined cost or expected return hits a target value exactly (or very closely).


## 2. Reduction Demonstration

We show a polynomial-time reduction from Subset Sum to Knapsack.

### Knapsack Problem (0/1 version) Definition

Given:
- A set of items, where each item i has:
  - weight wi
  - value vi
- A knapsack capacity C
- A required minimum value V

Question (decision version):

Is there a subset of items whose total weight is at most C and whose total value is at least V?

### Reduction Idea

Subset Sum can be seen as a special case of the Knapsack problem where:

- Each item's value is equal to its weight.
- The knapsack capacity is equal to the target T.
- The required value is also equal to T.

So:

- value_i = weight_i = xi for each element xi in S
- Capacity C = T
- Required value V = T

### Mapping / Transformation

Given a Subset Sum instance:

- Input:
  - A set S = { x1, x2, ..., xn }
  - A target T

We construct a Knapsack instance as follows:

- For each element xi in S, create an item i with:
  - weight wi = xi
  - value vi = xi
- Set knapsack capacity C = T
- Set required value V = T

This transformation is clearly polynomial-time, since we only create one item per element and copy numerical values.

### Correctness of the Reduction

- If there exists a subset S' ⊆ S such that sum_{x in S'} x = T,  
  then we can pick the corresponding items in the Knapsack instance.  
  Their total weight will be T and their total value will also be T, satisfying:
  - total weight ≤ C (because T ≤ T)
  - total value ≥ V (because T ≥ T)

- Conversely, if there is a subset of items in the Knapsack instance with:
  - total weight ≤ C = T, and
  - total value ≥ V = T,
  
  and since each item's weight and value are equal (wi = vi = xi),  
  the only way to have total value at least T and total weight at most T  
  is for the total weight and total value to both be exactly T.  
  Therefore, the corresponding subset of numbers in S sums exactly to T.

Thus, the Subset Sum instance has a solution if and only if the constructed Knapsack instance has a solution.

- The reduction runs in polynomial time.
- This shows that Knapsack is at least as hard as Subset Sum.


## 3. Algorithm Implementation in Python

We implement a Dynamic Programming solution for the Subset Sum decision problem.

Time complexity: O(n * T),  
where:
- n = number of elements in S
- T = target sum

Here is example Python code:


```python
# Subset Sum using Dynamic Programming
# Returns True if a subset sums to target T

def subset_sum(nums, T):
    n = len(nums)

    # DP table: (n+1) x (T+1)
    dp = [[False] * (T + 1) for _ in range(n + 1)]

    # A sum of zero is always possible (empty subset)
    for i in range(n + 1):
        dp[i][0] = True

    # Fill dp table
    for i in range(1, n + 1):
        for s in range(1, T + 1):
            # Option 1: don't take nums[i-1]
            dp[i][s] = dp[i-1][s]

            # Option 2: take nums[i-1] if it's not larger than current sum
            if nums[i-1] <= s:
                dp[i][s] = dp[i][s] or dp[i-1][s - nums[i-1]]

    return dp[n][T]


# Sample run
nums = [3, 34, 4, 12, 5, 2]
T = 9

print("Input set:", nums)
print("Target:", T)
print("Subset Sum exists?", subset_sum(nums, T))
```
