## **1.1 Introduction to Dynamic Programming**

Dynamic Programming (DP) is a problem-solving technique used when a problem can be broken down into **overlapping subproblems** whose results are reused many times. Instead of recomputing the same results repeatedly, DP **stores solutions** to smaller subproblems in a table (usually a list or matrix). This makes the algorithm far more efficient.

DP is particularly useful in optimization problems such as shortest paths, knapsack, Fibonacci, and Longest Common Subsequence (LCS). The key idea is:
**Solve small subproblems once then store results and finally build up to the final solution.**


## **1.2 Dynamic Programming Approach for LCS**

The LCS problem fits Dynamic Programming because - 

* It has **optimal substructure**: the LCS of two strings depends on the LCS of their prefixes.
* It has **overlapping subproblems**: the same LCS subproblems appear multiple times.

We create a 2D table `dp` where each cell `dp[i][j]` represents -

> **The LCS length of the first *i* characters of s1 and the first *j* characters of s2.**

We then fill the table using two rules:

### **Case 1 Characters Match**

If
`s1[i−1] == s2[j−1]`,
then that character contributes to the LCS:

```
dp[i][j] = dp[i−1][j−1] + 1
```

### **Case 2 Characters Do NOT Match**

We choose the best LCS obtained by dropping one character at a time:

```
dp[i][j] = max(dp[i−1][j], dp[i][j−1])
```

After filling the entire matrix, the final answer is stored in:

```
dp[len(s1)][len(s2)]
```

which represents the LCS length of the full strings.



## **2 LCS Code Implementation**

````python
def lcs(s1: str, s2: str) -> int:
    # Lengths of the strings
    m, n = len(s1), len(s2)

    # Create a DP table with (m+1) rows and (n+1) columns
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    # Fill the DP table
    for i in range(1, m + 1):
        for j in range(1, n + 1):

            # If characters match, extend the subsequence
            if s1[i - 1] == s2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            
            # Otherwise take the best of ignoring one character
            else:
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])

    # The bottom-right cell contains the LCS length
    return dp[m][n]
````
