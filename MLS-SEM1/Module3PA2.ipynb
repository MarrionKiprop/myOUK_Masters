# Python Implementation

````python
def insertion_sort(lt):
    n = len(lt)
    for k in range(1, n):
        temp = lt[k]
        l = k - 1
        # Shift larger elements one position to the right
        while l >= 0 and lt[l] > temp:
            lt[l + 1] = lt[l]
            l -= 1
        lt[l + 1] = temp
    return lt

# Test
print(insertion_sort([6, 7, 9, 2, 3, 1, 0]))
````

# Time Complexity

For the Worst Case will have time complexity of (O(n²)) - This takes quadratic time when the list is completely reversed and will require maximum shifting. The Average Case still needs (Θ(n²)) time complexity. On average, about half the elements get shifted each pass of the outer loop. Best case where all the items in the list are already sorted, it will take linear time Ω(n). In  this case no shifts are needed.
