### 1.0 AVERAGE CALCULATOR ALGORITHMN

In this activity, we consider a basic algorithm known as the Average Calculator. The goal of the algorithm is to compute the arithmetic mean of a set of numbers by summing all the values and then dividing the total by the number of elements. Although the task is simple, it provides a useful example for practising how to represent an algorithm clearly using both a flowchart and pseudocode before writing Python code. The flowchart presents the logic visually—showing the sequence of steps for inputting numbers, updating the running sum and count, and computing and displaying the final average—while the pseudocode expresses the same logic in a structured, language-independent form. ### 1.2 PSEUDOCODE
```
Algorithm AVERAGE_CALCULATOR
Input: An array A of n numbers
Output: A float (or integer if result is whole number)

1. set total ← 0                          # accumulator for sum
2. set count ← 0                          # number of elements processed
3. for i ← 0 to n - 1 do                  # iterate through each element
       total ← total + A[i]               # add current element to total
       count ← count + 1                  # increment counter
4. average ← total / count                # compute mean
5. return average
```

### 1.4 AVERAGE CALCULATOR CODE IMPLEMENTATION

````python

def average_calc(arr:tuple[int,...]) -> float:
    total = 0 
    count = 0
    for num in arr: # Loop over the array A
        total += num  # add num to total
        count += 1    # icrement counter
    return total/count # calculate average

A = 1,2,3,4,5,6,3,4,4,66,4,4,4,5,6 # Test array A
B = 2,3,2,2,2,4,2,4,2,4,5,6,7,3  # Test array B
C = 10,20,30,40,50,60,70,80,90

print(f"The Average is: {average_calc(A):.2f}")
print(f"The Average is: {average_calc(B):.2f}")
print(f"The Average is: {average_calc(C):.2f}")

````
