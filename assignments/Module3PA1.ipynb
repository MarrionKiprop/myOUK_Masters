# Linear Search Algorithm

### My PseudoCode Implementation -  Linear Search

```
1.  Set n ← length(A)
2.  For i ← 0 to n − 1 do
3.      If A[i] == k then
4.          Return True
5.  Return False
```

### Python Implementation of My Pseudocode

````python
def linear_search(arr, k):
    """ Do a Linear Search on an Array and return 
    True if element k is found or False if else """
    for se in range(len(arr)): # Loop through a lenght of the array arr
        if arr[se] == k: # Check if the element k is n the array
            return True
    return False
````

````python
import time # Importing time to measure the time it takes to run the algorithm
# First i am testing with 10 elements in a list
arr = [1, 3, 5, 6, 3, 4, 6, 4, 6, 7]
k = 7
start = time.time()
result_1 = linear_search(arr,k)
end = time.time()

print("Search Result:", result_1)
print(f"Time Taken: {end - start:.10f} seconds")
````

# Bubble Sort Algorithm

## My Pseudocode for the bubble sort algorithmn

```
1. Set n ← length(lst)

2. For i ← 0 to n - 1 do
      2.1 For j ← 0 to n - i - 2 do
             If lst[j] > lst[j + 1] then
                  temp ← lst[j]
                  lst[j] ← lst[j + 1]
                  lst[j + 1] ← temp
3. Return lst

```

## My Python Implementation of the above Bubble Sort Pseudocode

````python
def bubble_sort(lst):
    """ Code for a buble sort algorithmn using a nested 
    loop with comparison of elements with the outer loop 
    being the counting number of passes """
    n = len(lst)
    for i in range(n):
        for j in range(0, n - i - 1):
            if lst[j] > lst[j + 1]:
                lst[j], lst[j+1] = lst[j+1], lst[j] # Swap the elements j and j + 1
    return lst # Return the sorted list 
````

##  Testing the Bubble Sort Algorithmn with 10 elements

````python
import time # Importing time to measure the time it takes to run the algorithm
# First i am testing with 10 elements in a list
lst = [9,2,8,5,7,4,1,6,3,0]
start = time.time()
mysortedlst = bubble_sort(lst)
end = time.time()

print(f"Time Taken: {end - start:.10f} seconds")
````
