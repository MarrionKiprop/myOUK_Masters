### HTML Display of the Images

```python
import base64
import os
from IPython.display import HTML

def embed_image(path, width=250):
    """Convert an image to a self-contained HTML tag."""
    # FIX: os.path.splitext returns a tuple (root, ext). We select index 1.
    ext = os.path.splitext(path)[1].lower().replace('.', '')
    mime_type = "jpeg" if ext in ["jpg", "jpeg"] else ext
    
    with open(path, "rb") as f:
        data = base64.b64encode(f.read()).decode("utf-8")
    return f'<img src="data:image/{mime_type};base64,{data}" width="{width}" style="margin:5px;"/>'

# 1. Define your folder path exactly where the images live
folder_path = "/home/langstkip/workspace/myOUK_Masters/MLS-SEM3/CSA803/Images"

# 2. Grab all image files from that folder using the unified folder_path variable
valid_extensions = (".jpg", ".jpeg", ".png", ".webp")
image_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.lower().endswith(valid_extensions)]

# 3. Build and display the gallery
html_gallery = "<div style='text-align:center;'>"
for path in image_files:
    html_gallery += embed_image(path, width=250)
html_gallery += "</div>"

HTML(html_gallery)
```
