# Part A
# NLP Conceptual Questions

## Importance of Preprocessing in NLP Tasks

Text preprocessing is important in Natural Language Processing because raw text is usually noisy, inconsistent and difficult for a machine learning model to interpret directly. In real-world applications, text may contain different languages, informal comments, spelling mistakes, HTML tags, emojis, punctuation, numbers and repeated words. If this raw text is passed directly into an NLP model, the model may learn irrelevant patterns instead of meaningful language features. Preprocessing converts messy text into a cleaner and more structured form. This improves the quality of the input data and helps the model perform better in tasks such as classification, sentiment analysis, topic modelling and information retrieval.

## Tokenization in NLP Models

Tokenization helps NLP models by breaking large blocks of text into smaller units that can be processed and analysed. A sentence such as “NewsFlash collects articles daily” can be split into words such as “NewsFlash,” “collects,” “articles” and “daily.” These units become the basic features that the model uses to understand patterns in language. Common types of tokenization include word tokenization, where text is split into individual words, sentence tokenization, where text is split into sentences, and subword tokenization, where words are broken into smaller meaningful parts. Subword tokenization is useful for handling rare words, spelling variations and complex languages.

## Stemming and Lemmatization

Stemming and lemmatization are both used to reduce words to a simpler form, but they work differently. Stemming cuts a word down to its root form, sometimes without considering grammar. For example, “running,” “runs” and “runner” may be reduced to “run” or a rough stem such as “runn.” Lemmatization is more accurate because it uses vocabulary and grammar to convert a word into its proper dictionary form. For example, “better” can be lemmatized to “good,” and “running” can become “run.” Stemming is usually faster, while lemmatization is more linguistically correct and useful when meaning is important.

## Importance of Stop Word Removal

Stopword removal is important because some words appear very frequently but may add little value in certain NLP tasks. Words such as “the,” “is,” “and,” “to” and “in” can dominate the text without helping the model understand the main topic. Removing stopwords reduces the size of the data and allows the model to focus on more meaningful words. For example, in topic classification, words like “climate,” “policy,” “market” or “agriculture” may be more useful than common function words. However, stopword removal should be applied carefully because some stopwords may carry meaning in tasks such as sentiment analysis or question answering.

## Challenges in Non-English and Domain-specific Text

Non-English and domain-specific text creates additional preprocessing challenges. Different languages have different grammar, word order, morphology and punctuation rules, so a preprocessing method designed for English may not work well for Kiswahili, French or other languages. Some languages also require special tokenizers because words may be joined, inflected or written differently. Domain-specific text also has unique vocabulary, abbreviations and technical terms. For example, agricultural, medical or legal text may contain terms that general NLP tools do not understand. Therefore, preprocessing should be adapted to the language and domain. This may require language detection, custom dictionaries, domain-specific stopword lists and specialized tokenizers. When preprocessing is adapted properly, NLP models receive cleaner, more meaningful and context-aware input, which improves accuracy and reliability.


# Part B
# Practical NLP Python Implementation

## Objective

The objective of this section is to implement basic text preprocessing in Python using the NLTK library.

## Tasks 

1. Convert text to lowercase  
2. Remove noise such as punctuation and numbers  
3. Remove stopwords  
4. Apply stemming and lemmatization  


```python
# Import required libraries

import re
import string
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required NLTK resources.
# These downloads are needed only the first time the notebook is run.

resources = {
    "punkt": "tokenizers/punkt",
    "punkt_tab": "tokenizers/punkt_tab",
    "stopwords": "corpora/stopwords",
    "wordnet": "corpora/wordnet",
    "omw-1.4": "corpora/omw-1.4"
}

for resource, path in resources.items():
    try:
        nltk.data.find(path)
    except LookupError:
        nltk.download(resource)
```

```python
# Sample raw text representing noisy news and user comment data

raw_text = """
BREAKING NEWS!!! NewsFlash collected 1,000,000+ articles today from different websites.
<html><body>Climate change is affecting farmers in Kenya, and prices are rising fast!!!</body></html>
Users commented: "This is sooo shocking 😢😢 visit https://newsflash.example.com for more details."
AI-driven agriculture, food-security, and market updates were also discussed in 2026.
"""

print("Original Raw Text:")
print(raw_text)
```

```python
# Step 1: Convert text to lowercase
# Lowercasing ensures that words like "News", "NEWS" and "news" are treated as the same word.

lower_text = raw_text.lower()

print("Lowercase Text:")
print(lower_text)
```

```python
# Step 2: Remove noise
# This includes HTML tags, URLs, punctuation, numbers and extra spaces.

# Remove HTML tags
clean_text = re.sub(r"<.*?>", " ", lower_text)

# Remove URLs
clean_text = re.sub(r"http\S+|www\S+", " ", clean_text)

# Remove numbers
clean_text = re.sub(r"\d+", " ", clean_text)

# Remove punctuation
clean_text = clean_text.translate(str.maketrans("", "", string.punctuation))

# Remove emojis and other non-alphabetic symbols
clean_text = re.sub(r"[^a-zA-Z\s]", " ", clean_text)

# Remove extra spaces
clean_text = re.sub(r"\s+", " ", clean_text).strip()

print("Cleaned Text:")
print(clean_text)
```

```python
# Step 3: Tokenization
# Tokenization breaks the cleaned text into individual words.

tokens = word_tokenize(clean_text)

print("Tokens:")
print(tokens)
```

```python
# Step 4: Stopword removal
# Stopwords are common words such as "the", "is", "and" and "to".
# Removing them helps the model focus on more meaningful terms.

stop_words = set(stopwords.words("english"))

filtered_tokens = [word for word in tokens if word not in stop_words]

print("Tokens After Stopword Removal:")
print(filtered_tokens)
```

```python
# Step 5: Stemming
# Stemming reduces words to a root form.
# It is fast but may sometimes produce rough word forms.

stemmer = PorterStemmer()

stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]

print("Stemmed Tokens:")
print(stemmed_tokens)
```

```python
# Step 6: Lemmatization
# Lemmatization reduces words to their correct dictionary form.
# It is usually more meaningful than stemming.

lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]

print("Lemmatized Tokens:")
print(lemmatized_tokens)
```

```python
# Step 7: Compare preprocessing outputs

print("Original Text:")
print(raw_text)

print("\nCleaned Text:")
print(clean_text)

print("\nTokens:")
print(tokens)

print("\nAfter Stopword Removal:")
print(filtered_tokens)

print("\nAfter Stemming:")
print(stemmed_tokens)

print("\nAfter Lemmatization:")
print(lemmatized_tokens)
```

# Results

The preprocessing pipeline cleaned the raw text by converting it into lowercase, removing HTML tags, URLs, punctuation, numbers, emojis and extra spaces. The cleaned text was then tokenized into individual words. Stopwords were removed so that the remaining words carried more useful meaning for NLP analysis.

Stemming reduced words into shorter root forms, while lemmatization converted words into more meaningful dictionary forms. These preprocessing steps improve the quality of text data and make it easier for machine learning models to learn useful language patterns.
