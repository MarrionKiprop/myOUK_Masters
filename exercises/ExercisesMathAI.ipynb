```python
from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd

#Data Loading
r_data = pd.ExcelFile("rainfall_data.xlsx")
df = pd.read_excel("rainfall_data.xlsx",'Monthly_Rainfall')

df.set_index('Region', inplace = True)

#Set up Scaler and Standardize
scaler = StandardScaler()
df_x = scaler.fit_transform(df)

df_centered = df_x - np.mean(df_x)

df_cov = np.cov(df_centered.T)

#Calculate Eigens and sort
eigenvalues, eigenvectors = np.linalg.eig(df_cov)

idx = np.argsort(eigenvalues)[::-1]
s_eigenvalues = eigenvalues[idx]

s_eigenvectors = eigenvectors[:,idx]

var = np.real(s_eigenvalues / np.sum(s_eigenvalues)*100)

with np.printoptions(suppress=True, precision=2): # Rounded to 2 decimal places
    print(var)

for i , (eigenvalue, eigenvector, var_percentage) in enumerate(zip(s_eigenvalues[0:4], s_eigenvectors.T, np.real(var))):
   print("eigenvalue is:",np.round(eigenvalue, 2), 'eigenvector is:', np.round(eigenvector, 2), "variance percentageis :", np.round(var, 2) )

#Calculate for 90%
instance_x = 0
t_sum = 0
for i in np.real(var):  
    t_sum += i
    instance_x += 1 
    if t_sum >= 90:
        break
print(t_sum)
print(instance_x)
```

```python
def find_substring(text, pattern):
    n = len(text)
    m = len(pattern)

    for i in range(n - m + 1):
        sub = text[i:i + m]
        if sub == pattern:
            return i
    return -1

# Test
text = "this is a simple text"
pattern = "simple"
print(find_substring(text, pattern))  # Output: 10
```

```python
def find_second_largest(A):
    """
    Finds the second largest value in a list A.
    
    Parameters:
        A (list of numbers): The list to search.
    
    Returns:
        number: The second largest value in the list, or None if not found.
    """
    if len(A) < 2:
        return None  # Not enough elements

    max_val = second_max = float('-inf')

    for num in A:
        if num > max_val:
            second_max = max_val
            max_val = num
        elif num > second_max and num != max_val:
            second_max = num

    return second_max if second_max != float('-inf') else None

# Example usage
A = [3, 8, 2, 10, 6]
print("Second largest value is:", find_second_largest(A))
```

```python
def zero_sum(arr,target = 0):
    sum_z = []
    for ar in range(len(arr)):
        for ar2 in range(ar + 1, len(arr)):
            for ar3 in range(ar2 + 1, len(arr)):
                if arr[ar] + arr[ar2] + arr[ar3] == target:
                    y = sorted([arr[ar], arr[ar2], arr[ar3]])
                    if y not in sum_z:
                        sum_z.append(y)

    return sum_z
arr = [-1, 0, 1, 2, -1, -4]
zero_sum(arr, 0)
```

```python
def max_num(L):
    if len(set(L)) < 2:
         return "No Second Max"
    max_n = L[0]
    prev_max = float('-inf')
    for i in L:
        if  i > max_n:
            prev_max = max_n
            max_n = i
        if i > prev_max and i < max_n:
             prev_max = i
    return prev_max, max_n

L = [2,4,6,3,5,3,5,6,3,4,10]
max_num(L)
```

```python
def merge(left, right):
    result = []
    i = j = 0

    # Compare elements and merge in order
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Add any remaining elements
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)
arr = [10, 20, 30, 40, 50, 60]
merge_sort(arr)
```

```python
def counting_sort(arr):
    count = [0] * (max(arr) + 1)
    for num in arr:
        count[num] += 1
    sorted_arr = []
    for value, freq in enumerate(count):
        sorted_arr.extend([value] * freq)

    return sorted_arr
arr = [1,2,3,5,6,7,8,3,5,7,8,12,4,5,67]
counting_sort(arr)
```

```python
def read_file_to_dict(filename):
    student_dict = {}
    with open(filename, 'r', encoding="utf-8") as f:
        for line in f:
            if ":" in line:
                key, value = line.strip().split(":", 1)
                student_dict[key.strip()] = [float(value.strip())]
    return student_dict

filename = "students.txt"
students = read_file_to_dict(filename)

def calculate_average(student_dict):
    average = 0
    count = 0
    for grades in student_dict.values():
        average += sum(grades)
        count += 1
    return average/count
averages  = calculate_average(students)
print(averages)
def print_above_average_students(student_dict, averages):
    above_average = {}
    for name, grades in student_dict.items():
        if grades[0] > averages:
            above_average[name] = grades[0]
    return above_average
l = print_above_average_students(students, averages)
print(l)
```

```python
import numpy as np

def reconstruct_matrix(A):
    
    # Compute eigenvalues and eigenvectors
    eigenvalues, eigenvectors = np.linalg.eig(A)

    # Form the diagonal matrix of eigenvalues
    Λ = np.diag(eigenvalues)

    # Compute the inverse of the matrix of eigenvectors
    Q_Inv = np.linalg.inv(eigenvectors)

    # Reconstruct the matrix
    A_reconstructed = np.dot(np.dot(eigenvectors, Λ), Q_Inv)
    
    return A_reconstructed, eigenvalues, eigenvectors

A = np.array([[3,1],[0,2]])

print(reconstruct_matrix(A))
```
