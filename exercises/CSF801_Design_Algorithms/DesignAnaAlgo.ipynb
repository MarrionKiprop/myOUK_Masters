```python
def is_prime_n(a):
    for i in range(a-1,1,-1):
        if a % i == 0:
            return False
    return True
is_prime_n(59)
```

```python
def is_factors(a):
    factors = []
    for i in range(a,0,-1):
        if a % i == 0:
            factors.append(i)
    return factors
is_factors(36)
```

```python
def sum_digits(n):
    tsum = 0
    while n > 0:
        digits = n % 10
        tsum += digits
        n = n // 10
    return tsum
sum_digits(3459)
```

```python
def rev(n):
    num = 0
    while n > 0:
        digit = n % 10
        num = num * 10 + digit
        n = n // 10 
    return num
rev(345)
```

```python
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    
    a, b = 0, 1
    for i in range(2, n + 1):
        next = a + b
        a = b
        b = next
    return b
fib(14)
```

```python
def odd_digits(n):
    while n > 0:
        digit = n % 10
        if digit % 2 != 0:
            yield digit
        n = n // 10
print(list(odd_digits(4566)))
```

```python
def odd_even_digits(n):
    odd1 = []
    even1 = []
    while n > 0:
        digit = n % 10
        if digit % 2 != 0:
            odd1.append(digit)
        else:
            even1.append(digit)
        n = n // 10
    return odd1, even1
odd_even_digits(4567)
```

```python
def has_pair_fast(arr, target):
    seen = set()
    for num in arr:
        if target - num in seen:
            return True
        seen.add(num)
arr = [1,2,4,5,3,4,5,6,5]
has_pair_fast(arr,20)
```
