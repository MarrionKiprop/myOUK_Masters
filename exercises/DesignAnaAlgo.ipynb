```python
def average_calc(arr:tuple[int,...]) -> float:
    total = 0 
    count = 0
    for num in arr: # Loop over the array A
        total += num  # add num to total
        count += 1    # icrement counter
    return total/count # calculate average

A = 1,2,3,4,5,6,3,4,4,66,4,4,4,5,6 # Test array A
print(f" The Average is, {average_calc(A)}") 
```

Algorithm AVERAGE_CALCULATOR
Input: An array A of n numbers
Output: A float (or integer if result is whole number)

1. set total ← 0                          # accumulator for sum
2. set count ← 0                          # number of elements processed
3. for i ← 0 to n - 1 do                  # iterate through each element
       total ← total + A[i]               # add current element to total
       count ← count + 1                  # increment counter
4. average ← total / count                # compute mean
5. return average





```python
data =  {
  "CowA": [5, 6, 7, 5, 6, 7, 8],
  "CowB": [4, 5, 4, 5, 6, 5, 6],
  "CowC": [7, 8, 7, 8, 9, 8, 9]
}
 
def analyze_milk(data):
    results = {}
    for cow, yields in data.items():
        total = sum(yields)
        average = total / len(yields)
        results[cow] = {"total": total, "average": average}
    top_cow = max(results, key=lambda c: results[c]["total"])
    return results, top_cow
print(analyze_milk(data))
```

```python
import heapq

def dijkstra(graph, start, end):
    queue = [(0, start, [])]
    visited = set()
    while queue:
        (cost, current, path) = heapq.heappop(queue)
        if current in visited:
            continue
        visited.add(current)
        path = path + [current]
        if current == end:
            return (cost, path)
        for neighbor, weight in graph.get(current, []):
            if neighbor not in visited:
                heapq.heappush(queue, (cost + weight, neighbor, path))
    return float("inf"), []

# Graph definition
graph = {
    'A': [('B', 4), ('C', 2)],
    'B': [('C', 5), ('D', 10)],
    'C': [('D', 3), ('E', 1)],
    'D': [('E', 7), ('G', 3), ('F', 6)],
    'E': [('G', 4), ('F', 5), ('H', 5)],
    'F': [('G', 2), ('H', 3), ('I', 4)],
    'G': [('H', 2), ('I', 3), ('J', 4)],
    'H': [('I', 1), ('J', 2), ('K', 5)],
    'I': [('J', 2), ('K', 3), ('L', 4)],
    'J': [('K', 1), ('L', 2), ('M', 3)],
    'K': [],                     
    'L': [('M', 2), ('N', 3), ('O', 4)],
    'M': [],
    'N': [],
    'O': []
}


# Execution
start, end = 'A', 'F'
distance, path = dijkstra(graph, start, end)
print(f"Shortest Path: {' -> '.join(path)}")
print(f"Total Distance: {distance}")
```

```python
def uniquep(m,n):
    row = [1] * n

    for i in range(m-1):
        newRow = [1] * n
        for j in range(n - 2, -1, -1):
            newRow[j] = newRow[j + 1] + row[j]
        row = newRow
    return row[0]

```
