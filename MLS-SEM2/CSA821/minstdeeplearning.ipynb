### Deep Learning Implementation with TensorFlow

The MNIST dataset is a widely used benchmark for handwritten digit recognition. Each image is a 28×28 grayscale matrix, giving a total of 784 features per sample. Compared to modern image datasets, MNIST is relatively small in both size and complexity, but it still requires models that can handle high-dimensional data effectively.

For this assignment, TensorFlow ws chosen rather than scikit-learn. While scikit-learn is excellent for classical machine learning methods such as logistic regression, decision trees, and PCA, it treats the pixel values as independent features. This approach does not fully capture the spatial structure of the images. TensorFlow, on the other hand, is built around tensor operations, making it naturally suited for image data. Neural networks in TensorFlow can learn complex, non-linear relationships between pixels and exploit the structure of the data more effectively.

By using TensorFlow’s integrated Keras API, we were able to design a simple yet powerful neural network architecture that achieved high accuracy on MNIST. This demonstrates the strength of deep learning methods for image recognition tasks, even when the dataset has fewer features compared to modern large-scale datasets.

```python
import tensorflow as tf
mnist = tf.keras.datasets.mnist

(x_train, y_train),(x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

model = tf.keras.models.Sequential([
  tf.keras.layers.Flatten(input_shape=(28, 28)),
  tf.keras.layers.Dense(128, activation='relu'),
  tf.keras.layers.Dropout(0.2),
  tf.keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
  loss='sparse_categorical_crossentropy',
  metrics=['accuracy'])

model.fit(x_train, y_train, epochs=5)
model.evaluate(x_test, y_test)
```

### Model Explanation

The model used in this assignment is a feedforward neural network built with TensorFlow’s Keras API. Its purpose is to classify handwritten digits (0–9) from the MNIST dataset. The architecture begins with a Flatten layer, which reshapes each 28×28 image into a 784-dimensional vector so that the dense layers can process the pixel values as features.

Next, a Dense hidden layer with 128 neurons and ReLU activation is applied. This layer enables the network to learn non-linear relationships between pixels, allowing it to capture complex patterns in the digit images. To improve generalization and reduce overfitting, a Dropout layer with a rate of 0.2 randomly deactivates neurons during training.

The final Dense output layer contains 10 neurons with softmax activation, producing a probability distribution across the ten digit classes. The digit with the highest probability is selected as the model’s prediction.

The model was compiled using the Adam optimizer, which is an adaptive gradient descent method, and the sparse categorical crossentropy loss function, which is suitable for integer labels. Accuracy was chosen as the evaluation metric. Training was carried out for five epochs on normalized data, where pixel values were scaled to the range [0,1] to ensure faster convergence and stable learning.

On evaluation with the test set, the model achieved a loss of 0.067 and an accuracy of 97.9%, demonstrating excellent performance in recognizing handwritten digits. This confirms that even a relatively simple feedforward neural network can achieve high accuracy on MNIST when implemented with TensorFlow.
